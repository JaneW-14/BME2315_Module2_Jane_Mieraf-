## Project Goal:
As a team our goal is to determine if there are physiological or functional measures that can be used to measure facioscapulohumeral muscular dystrophy (FSHD) progression, beyond a traditional severity score. If measures are found that are proven to be accurate measurements of disease progression, we argue they should be used in clinical trials as additional endpoints. These additional endpoints could increase the accuracy of clinical trials by providing additional chances for trial success. Optimistically, more medications will pass the clinical trial phase and will be able to be put into market. As a team we are specifically focused on comparing the sensitivity of fat fraction in the TA and the functionality scores in the 10 meter walk run test. 

Our question is: 
Which measure is more sensitive to disease progression: the fat fraction in the left and right TA (tibialis anterior) or the score on the 10 meter walk run test (WR)? Furthermore, can a change in fat fraction in the TA in the first 12 months predict the patient's performance in the 10 meter WR test over the course of 2 years? 

By examining the fat fraction score in conjunction with a functionality test we are examining which is a better predictor of disease severity, and if the two measures impact each other. These results would provide vital knowledge that could be applied to designing clinical trials. We are specially looking at fat fraction in the TA because it is an easily accessible muscle to biopsy and commonly damaged over the course of FSHD progression. Furthermore, after conducting preliminary analysis with AI engine Claude, the fat fraction percent increase from baseline on the left side was 12.7% and on the right side was 10.6%, the two largest increases across all the measures in the provided data set. Therefore, we know the change in fat fraction has a similar trend as change in disease severity. These results provide justification for investigating fat fraction specifically. 
The TA contributes to an individual's ability to lift their foot off the ground, therefore our hypothesis is that as the fat fraction in the TA increases, strength in the TA decreases and the walk run score will decrease as well. We are examining specifically the walk and run test because it is the last stage of mobility before a patient is required to use a wheelchair. Therefore, the walk run test has the longest timespan and therefore possible data to track disease progression. 


## Disease Background: 
FSHD Disease Background

### Prevalence & Incidence
FSHD is the third most common inherited muscular dystrophies. Estimated global prevalence ranges between 1 in 7,500 and 1 in 20,000 individuals (or about 4 to 12 cases per 100,000 individuals) (Deenen et al., 2014). In the United States, total prevalent cases are estimated to be between 75,000 and 80,000 affected individuals (FSHD Society, 2014). FSHD may be underestimated and milder cases can remain undiagnosed (Deenen et al., 2014). The annual incidence rate is estimated to be 1 in 20,000 live births (Medscape, 2019.). Different regional/population studies can have varied recorded prevalence rates. For example, in the Netherlands, estimates of prevalence have a higher rate, whereas regions like China report lower prevalence rates due to unequal diagnostic access and underdiagnosis (Deenen et al., 2014).

### Symptoms
Most symptoms manifest during adolescence of young adulthood, usually before twenty years of age. However, onset can span from early infancy to later adult life. The clinical feature of FSHD is an asymmetric pattern of progressive skeletal muscle weakness that begins in the upper body. The first sign is typically facial muscle weakness, which then shifts to weakness in the shoulder blades, scapular fixators, upper arms. The shoulder blades protrude outward to display “scapular winging.” Common complications include breathing problems, foot drop, lumbar lordosis from weak abdominal muscles, retinal damage, and hearing loss. These motor symptoms heavily impact the ability to complete routine daily tasks. Additionally, musculoskeletal pain and fatigue is common (Muscular Dystrophy Association, 2015.).

### Diagnosis
FSHD diagnosis begins with a clinical evaluation that examines muscle weakness, symptom onset, family history, and the distribution of affected muscles. Physicians look for physical signs such as asymmetric facial weakness, shoulder girdle atrophy, and scapular winging during physical examination. Genetic testing is the primary method used to confirm the diagnosis and distinguish FSHD type 1 from FSHD type 2. FSHD type 1, which accounts for approximately 95% of cases, is associated with contraction of the D4Z4 repeat region on chromosome 4, while FSHD type 2 (5% of cases) involves abnormal epigenetic regulation of this region. Additional tests, including creatine kinase blood testing, muscle MRI, and muscle biopsy may help evaluate muscle damage or rule out other conditions (FSHD Society, 2019). Muscle MRIs are used to detect early muscular inflammation and fatty replacement, while creatine kinase levels remain normal or slightly elevated. Biopsy is now done only when FSHD is suspected but genetic testing does not confirm it (Preston & Wang, 2025).

### Standard of Care Treatments (& reimbursement)
Currently, there are no FDA-approved disease-modifying therapies or medication to cure or halt the progression of FSHD, leaving clinical care strictly focused on supportive symptom management and improving quality of life. Multidisciplinary care teams emphasize physical therapy and individualized exercise programs to preserve existing muscle function, maintain joint range of motion, flexibility, strength, and prevent contractures. Orthotic devices such as wheelchairs or ankle-foot orthoses (AFOs) are regularly prescribed to assist patients experiencing foot drop and improve walking safety. Some patients can undergo scapulopexy (scapular fixation surgery), to stabilize the shoulder blades and improve arm reach and mobility. Managing chronic pain is a priority, often involving nonsteroidal anti-inflammatory drugs (NSAIDs) and neuropathic pain medications like gabapentin (FSHD Society, 2019; Preston & Wang, 2025). Insurance reimbursement for physical therapy, assistive devices, surgery, pain management, and specialized genetic testing depends on medical necessity, the patient's insurance plan, and applicable coverage policies. 

### Disease Progression & Prognosis
The natural progression of FSHD is typically slow, insidious, and variable, even among family members sharing the exact same genetic mutation. The severity can vary considerably between individuals as some symptoms may begin during childhood or adulthood. Some may experience mild weakness while others can develop significant physical disability. Muscle weakness generally follows a descending pattern, starting in the facial and shoulder muscles before gradually spreading downward toward the trunk, abdomen, and lower legs. Approximately 20% of affected individuals eventually require the use of a wheelchair or mobility scooter by age fifty to aid in performing everyday activities. The majority of individuals with classical FSHD have a normal life expectancy. However, if a patient presents with cardiac and respiratory complications or early-onset/infantile forms, they may require mechanical ventilation. Healthcare professionals adapt treatment throughout the patient's lifespan to meet each patient's needs (Muscular Dystrophy Association, 2026; MedlinePlus Genetics, 2025; Preston & Wang, 2025).

### Biological Mechanisms (anatomy, organ physiology, cell & molecular physiology)
FSHD is caused by abnormal regulation of the DUX4 gene, which is located within the D4Z4 repeat region near the end of chromosome 4. In FSHD type 1, contraction of the D4Z4 repeat region makes it easier for DUX4 to become abnormally active in skeletal muscle cells. In FSHD type 2, mutations in genes involved in epigenetic regulation, particularly SMCHD1, can also reduce the normal silencing of this region. When DUX4 is expressed in muscle cells, it disrupts normal cellular processes and contributes to cellular stress, apoptotic cascades, muscle-cell damage, immune system activation, inflammation-related responses, and progressive muscle degeneration and muscle fiber necrosis (MedlinePlus Genetics, 2025; Lemmers et al., 2015).

### Clinical Trials/Next-gen therapies
Treatments that reduce DUX4 expression could potentially address an underlying cause of the disease rather than simply managing its symptoms. One approach involves antisense oligonucleotides and other RNA-targeting technologies designed to reduce DUX4 messenger RNA and limit production of the harmful protein. Del-brax, also known as AOC 1020, is an experimental therapy to reduce DUX4 messenger RNA in muscle cells. Del-brax has been evaluated in clinical trials for FSHD (ClinicalTrials.gov, 2026). Another drug, losmapimod, was studied in a Phase 3 clinical trial, but it did not show a significant benefit compared with a placebo, so development for FSHD was discontinued in 2024 (Fulcrum Therapeutics, 2024). Other research focuses on improving drug delivery to skeletal muscle, identifying biomarkers that measure disease activity, and developing more sensitive methods for evaluating muscle function. Although these therapies show potential, more research is needed to determine whether they are safe and effective for patients.

# MRI Data Analysis Methdology:

### Data Loading and Preprocessing
The MRI analysis was performed using Python, using NumPy for image-array processing, pandas for organizing metadata and results, and Matplotlib for visualizing the images and fat fraction measurements. The dataset contains paired water and fat MRI cross-sections of the TA from one participant during two visits. Each visit has eight slices or cross-sectional locations along the TA, which are not equally spaced. The slice_info.csv file included each slice’s physical distance and percentage location along the muscle. A metadata table links each slice to its water image, fat image and binary TA segmentation mask.

```python
from pathlib import Path
import numpy as np
import ipykernel
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_TA_MRI_data_0020_TP1_TP2")
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")
manifest
row = manifest[manifest["timepoint"] == "TP1"].iloc[3]  # representative mid-muscle slice

water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```

### Fat-Fraction Calculation
Fat fraction was calculated for each pixel as fat/(water+fat) * 100. To avoid division by zero and complications in the code, fat fraction was calculated where the sum of water and fat had to be greater than 0. The TA segmentation mask was able to distinguish between muscle pixels and surrounding tissue, which was excluded from the muscle-specific measurements. 


```python
denominator = water + fat
valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100

# Keep the whole supplied image region for the first analysis.
fat_fraction_full = fat_fraction.copy()

# Then apply the TA mask for the muscle-specific analysis.
fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]

print(f"Whole supplied image region — mean: {np.mean(whole_image_values):.1f}%")
print(f"Whole supplied image region — minimum: {np.min(whole_image_values):.1f}%")
print(f"Whole supplied image region — maximum: {np.max(whole_image_values):.1f}%")
print()
print(f"TA muscle only — mean: {np.mean(muscle_values):.1f}%")
print(f"TA muscle only — minimum: {np.min(muscle_values):.1f}%")
print(f"TA muscle only — maximum: {np.max(muscle_values):.1f}%")
```

### Image Visualization and Distribution
The water and fat images were displayed with the TA mask boundary to show which pixels belonged to the muscle. Fat fraction maps were generated using the 0-100% scale. 


```python
# Find the bounding box of the TA mask
y_pixels, x_pixels = np.where(mask)

y0, y1 = y_pixels.min(), y_pixels.max() + 1
x0, x1 = x_pixels.min(), x_pixels.max() + 1

# Crop the TA-specific fat-fraction map
fat_fraction_TA = fat_fraction_masked[y0:y1, x0:x1]

fig, axes = plt.subplots(1, 4, figsize=(15, 4))

# Water image with TA mask outline
axes[0].imshow(water, cmap="gray")
axes[0].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[0].set_title("Water image + TA mask")

# Fat image with TA mask outline
axes[1].imshow(fat, cmap="gray")
axes[1].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[1].set_title("Fat image + TA mask")

# Fat fraction for the whole supplied image region
whole_image = axes[2].imshow(
    fat_fraction_full,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[2].contour(mask, levels=[0.5], colors="cyan", linewidths=2)
axes[2].set_title("Fat fraction: whole region")

# Zoomed-in fat fraction for the TA only
ta_image = axes[3].imshow(
    fat_fraction_TA,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[3].set_title("TA fat fraction")

for ax in axes:
    ax.axis("off")

# Both fat-fraction plots use the same 0–100% scale
fig.colorbar(
    ta_image,
    ax=axes[2:],
    fraction=0.046,
    pad=0.04,
    label="Fat fraction (%)"
)

plt.show()
```

### Summary Statistics across muscle locations
The mean, median, and standard deviation of TA fat fraction were calculated for each slice. Additionally, the number of valid muscle pixels. The purpose of these measurements is to describe average muscle fat fraction, central value, variability among muscle pixels, and size of the analyzed region. The analysis was repeated for every sampled location in one visit.


```python
def calculate_fat_fraction(water, fat, mask):
    """Return a masked fat-fraction image in percent."""
    denominator = water + fat
    result = np.divide(
        fat,
        denominator,
        out=np.zeros_like(fat, dtype=float),
        where=denominator > 0
    ) * 100
    result[~mask] = np.nan
    return result


analysis_timepoint = "TP1"  # analyze one visit at a time
analysis_manifest = manifest[manifest["timepoint"] == analysis_timepoint]
results = []


for _, row in analysis_manifest.iterrows():
    water = np.load(DATA_DIR / row["water_file"])
    fat = np.load(DATA_DIR / row["fat_file"])
    mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
    ff = calculate_fat_fraction(water, fat, mask)


    results.append({
        "subject": row["subject"],
        "timepoint": row["timepoint"],
        "slice_id": row["slice_id"],
        "percent_along_muscle": row["percent_along_muscle"],
        "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
        "mean_fat_fraction_percent": np.nanmean(ff),
        "median_fat_fraction_percent": np.nanmedian(ff),
        "std_fat_fraction_percent": np.nanstd(ff),
        "n_muscle_pixels": int(mask.sum()),
    })


results_df = pd.DataFrame(results)
results_df
```

### Comparison between visits
To examine how fat fraction varied along the muscle, the mean fat fraction was plotted against physical distance. Since slices were not equally spaced, the supplied distances were used.


```python
def summarize_visit(timepoint):
    visit_results = []
    visit_manifest = manifest[manifest["timepoint"] == timepoint]


    for _, row in visit_manifest.iterrows():
        water = np.load(DATA_DIR / row["water_file"])
        fat = np.load(DATA_DIR / row["fat_file"])
        mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
        ff = calculate_fat_fraction(water, fat, mask)


        visit_results.append({
            "timepoint": timepoint,
            "slice_id": row["slice_id"],
            "percent_along_muscle": row["percent_along_muscle"],
            "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
            "mean_fat_fraction_percent": np.nanmean(ff),
        })


    return pd.DataFrame(visit_results)


profiles_df = pd.concat(
    [summarize_visit("TP1"), summarize_visit("TP2")],
    ignore_index=True
)


plt.figure(figsize=(7, 4))
for timepoint, group in profiles_df.groupby("timepoint"):
    group = group.sort_values("distance_from_muscle_start_mm")
    plt.plot(
        group["percent_along_muscle"],
        group["mean_fat_fraction_percent"],
        marker="o",
        label=timepoint
    )
plt.xlabel("Distance from muscle start (mm)")
plt.ylabel("Mean TA fat fraction (%)")
plt.title("Mean fat fraction along the tibialis anterior")
plt.legend(title="Visit")
plt.grid(True, alpha=0.3)
plt.show()
```

Figure 3 Caption: Mean TA Fat Fraction versus measured distance for TP1 and TP2.

```python

# INTERPOLATION CODE 

# Define the target time points for interpolation by using the minimum lag time at each time point 
T1 = patients["lag_months_y1"].min()
T2 = patients["lag_months_y2"].min()

# Define the measurements to be interpolated and their corresponding baseline, T1, and T2 fields.
measurements = {
    "ff_R": ("ff_baseline_R", "ff_y1_R", "ff_y2_R"),
    "ff_L": ("ff_baseline_L", "ff_y1_L", "ff_y2_L"),
    "wr_10": ("wr_10_baseline", "wr_10_y1", "wr_10_y2"),
}

#create a function called interpolate_measurement that requires the rows, fields, target date to interpolate to and version of interpolation (linear or quadratic)
#this code was created by the built in Copilot 

def interpolate_measurement(row, fields, target, kind):
    baseline_field, y1_field, y2_field = fields

    times = np.array([
        0.0,
        row["lag_months_y1"],
        row["lag_months_y2"],
    ])

    values = np.array([
        row[baseline_field],
        row[y1_field],
        row[y2_field],
    ])

    point_count = 2 if kind == "linear" else 3

    return float(
        interp1d(
            times[:point_count],
            values[:point_count],
            kind=kind,
        )(target)
    )

#this creates a table to display the results of the interpolation 
interpolated_patients = pd.DataFrame({
    "subject_id": patients["subject_id"],
    "T1_months": T1,
    "T2_months": T2,
})

for measure, fields in measurements.items():
    #for interpolating data to the t1 marker the code uses a linear interpolation because there are only two data points 
    interpolated_patients[f"{measure}_T1"] = patients.apply(
        interpolate_measurement,
        axis=1,
        fields=fields,
        target=T1,
        kind="linear",
    )
    #for interpolating data to the t2 marker the code uses a quadratic interpolation because there are three data points
    interpolated_patients[f"{measure}_T2"] = patients.apply(
        interpolate_measurement,
        axis=1,
        fields=fields,
        target=T2,
        kind="quadratic",
    )


# Add original baseline measurements to the data frame to allow them to be plotted alongside the interpolated values
interpolated_patients["ff_R_baseline"] = patients["ff_baseline_R"]
interpolated_patients["ff_L_baseline"] = patients["ff_baseline_L"]
interpolated_patients["wr_10_baseline"] = patients["wr_10_baseline"]

#this exports the interpolated data to a CSV file
interpolated_patients.to_csv(interpolated_filename, index=False)
```

## Interpolation Analysis:
Interpolation allows us to compare data that was not collected at the same time. This is especially crticial for our research question because our data analysis test Standardardized Response Mean looks at the change between data collection points. The study was desgined to test patients at 0 months, 12 months and 24 months, however this was not achived because every patient has thier own life that interfers. Therefore, interpolation is required to create data points that can be compared. These new data points have the same lag value (time between current test and previous test). This "target time point" is determined by choosing the shortest amount of time between the timepoints from the actual data set. This was done in the T1 = patients["lag_months_y1"].min() line of code above. By choosing a minimum value we avoid having to extrapolate data because we are certain all the data points fall within our set range. The minimum value for T1, first time point, was 11.5. The minimum value for T2, the second testing time, was 22.13. Therefore T1 = 11.15 months; T2 = 22.13 months and these will stay constant for the rest of our analysis.

Using these target time points we can interpolate all of the data from each patient, as if it was collceted at T1 and T2. To interpolate data taken at T1, we use a linear model because there are two data points that can be considered when making the estimation. To interpolate data taken at T2 we can use a quadratic model because there are three data points that can be considered when making the estimation. 

Uncertaintity: 

An interpolation is conducted for every individual patient, therefore, depending on how far from the target point (T1 = 11.5 and T2 = 22.13) the data is, the validity of the interpolation estimate changes. Patient data that was collected closer to the target T1 and T2 time frame will have a more accurate interpolation. 

```python

# Plot measured visits and interpolated values
fig, axes = plt.subplots(3, 1, figsize=(10, 12), sharex=True)

plot_measurements = [
    ("ff_R", "Right fat fraction (%)"),
    ("ff_L", "Left fat fraction (%)"),
    ("wr_10", "WR-10"),
]

for ax, (measure, ylabel) in zip(axes, plot_measurements):
    baseline_field, y1_field, y2_field = measurements[measure]

    for patient_index, (_, patient) in enumerate(patients.iterrows()):
        visit_times = [
            0,
            patient["lag_months_y1"],
            patient["lag_months_y2"],
        ]

        visit_values = [
            patient[baseline_field],
            patient[y1_field],
            patient[y2_field],
        ]

        ax.plot(
            visit_times,
            visit_values,
            color="gray",
            alpha=0.2,
            linewidth=0.8,
            label="Measured visits" if patient_index == 0 else None,
        )

        ax.scatter(
            visit_times,
            visit_values,
            color="black",
            alpha=0.45,
            s=18,
            zorder=3,
        )

    ax.scatter(
        np.full(len(patients), T1),
        interpolated_patients[f"{measure}_T1"],
        color="tab:blue",
        marker="D",
        alpha=0.75,
        s=32,
        label=f"Linear estimate at T1 ({T1:.2f} months)",
        zorder=4,
    )

    ax.scatter(
        np.full(len(patients), T2),
        interpolated_patients[f"{measure}_T2"],
        color="tab:orange",
        marker="D",
        alpha=0.75,
        s=32,
        label=f"Quadratic estimate at T2 ({T2:.2f} months)",
        zorder=4,
    )

    ax.axvline(T1, color="tab:blue", linestyle=":", alpha=0.6)
    ax.axvline(T2, color="tab:orange", linestyle=":", alpha=0.6)

    ax.set_ylabel(ylabel)
    ax.grid(True, alpha=0.25)
    ax.legend()


axes[-1].set_xlabel("Months since baseline")

fig.suptitle("Measured Visits and Interpolated Values for All Patients")
fig.tight_layout()
fig.savefig(graph_filename, dpi=200, bbox_inches="tight")

# Close the figure after saving so the script does not wait for a plot window.
plt.close(fig)

print(f"\n Interpolation: T1 = {T1:.2f} months; T2 = {T2:.2f} months")
print(f"Interpolated {len(patients)} patients: {interpolated_filename}")
print(f"Graph saved: {graph_filename}")
```

## T-Test 

To analyze the change in fat fraction we needed to combine the fat fraction measurements in the left and right TA. However, to make sure there is no difference between the left and right side that would prevent us from being able to combine them, we conducted a T-test. The T-test evaluates the significance of the differences between different groups. It outputs a value for p. If p>0.05 there is no significant difference in the two groups. If p<0.05 there is signifance, and we would not be able to combine them without misrepresentng the data. 

```python

# T-TEST CODE ------------------
# Perform paired t-tests to compare left and right fat fraction at each visit.

# Compare left and right fat fraction at each visit
fat_fraction_visits = {
    "Baseline": ("ff_L_baseline", "ff_R_baseline"),
    "T1": ("ff_L_T1", "ff_R_T1"),
    "T2": ("ff_L_T2", "ff_R_T2"),
}

test_results = []

for visit, (left_col, right_col) in fat_fraction_visits.items():
    paired_data = interpolated_patients[
        [left_col, right_col]
    ].dropna()

    left = paired_data[left_col]
    right = paired_data[right_col]

    # Safeguard: a paired t-test needs at least two complete pairs.
    if len(paired_data) < 2:
        t_stat, p_value = np.nan, np.nan
    else:
        t_stat, p_value = stats.ttest_rel(left, right)

    test_results.append({
        "Visit": visit,
        "N paired": len(paired_data),
        "Mean left": left.mean(),
        "Mean right": right.mean(),
        "Mean difference (L-R)": (left - right).mean(),
        "t-statistic": t_stat,
        "p-value": p_value,
    })

    # Average both sides for each patient
    average_col = f"ff_TA_average_{visit}"

    interpolated_patients[average_col] = (
        interpolated_patients[[left_col, right_col]]
        .mean(axis=1, skipna=False)
    )


results_df = pd.DataFrame(test_results)

print("\nPaired t-tests: Left vs. Right TA Fat Fraction")
print(results_df.round(4).to_string(index=False))
```

Three paired t-tests were conducted to compare left and right TA muscle fat fraction at baseline, T1, and T2 in patients with FSHD. The first step in our code was to define the three visits and their columns. The baseline was the original left and right TA fat-fraction measurements. T1 and T2 were the interpolated left and right measurements at 11.15 and 22.13 months. Next, to ensure that the comparisons use complete pairs, the code loops through each visit, selects the paired measurements, and removes rows where either measurement is missing. At this point, each paired t-test was run by using stats.ttest_rel(left, right) from the SciPy library. For each patient, the code calculates the difference and tests whether the population’s mean paired difference is zero. A small safeguard checks that at least two complete patient pairs are available before running the test. For each visit, the code records the number of paired patients, mean left TA fat fraction, mean right TA fat fraction, mean difference, t-statistic, and p-value. The results are converted into a table and printed. 

The results show that the mean left-side fat fraction is higher at every visit. All three p-values are greater than the significance threshold of 0.05, which means that there is no statistical difference between left and right TA fat fraction at any of the three visits. This does not prove that the two sides are equivalent, however, the findings can help to guide the decision to average left and right TA fat-fraction measurements for future data analyses.

# Comparing the change in fat fraction from baseline-T1 to the change in WR-10 time from baseline-T2

In order to analyze whether fat fraction can predict functional test results we want to analyze if the change in fat fraction in a year, will predict the long term functional change in the patient. To do this we used the interpolated values found before and graphed FF Baseline to T1 against WR-10 from baseline to T2. 

The value of R sqaured can be interpreted as the correlation strength between FF and WR-10. The max value for R squared is 1. The closer to 1, the stronger the correlation. 

```python

#TESTING IF THE CHANGE IN FF IN 0-12 MONTHS CAN PREDICT THE CHANGE IN WR-10 FROM 0-24 MONTHS 

#assigning values to the x and y arrays for the regression analysis
x_values = np.array([difference[0] for difference in difference_baseline_T1])
y_values = np.array(change_baseline_T2_WR)
valid_values = np.isfinite(x_values) & np.isfinite(y_values)
x_values = x_values[valid_values]
y_values = y_values[valid_values]

#creating the scatter plot and regression line for the relationship between fat fraction change and WR-10 change
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(x_values, y_values, alpha=0.75)
ax.set_xlabel("Fat fraction change: baseline to T1")
ax.set_ylabel("WR-10 change: baseline to T2")
ax.set_title("Fat Fraction Change vs. WR-10 Change")
ax.grid(True, alpha=0.25)

#making sure there are enoguh valid data points to perform the regression analysis
if len(x_values) >= 2 and np.ptp(x_values) > 0:
    regression = stats.linregress(x_values, y_values)
    line_x = np.linspace(x_values.min(), x_values.max(), 100)
    ax.plot(
        line_x,
        regression.intercept + regression.slope * line_x,
        color="tab:red",
        label=f"Linear regression (R² = {regression.rvalue ** 2:.3f})",
    )
    ax.legend()
    print(
        "\nLinear regression for fat fraction change (baseline to T1) "
        "vs. WR-10 change (baseline to T2): "
        f"slope={regression.slope:.4f}, "
        f"intercept={regression.intercept:.4f}, "
        f"R²={regression.rvalue ** 2:.4f}, "
        f"p-value={regression.pvalue:.4g}"
    )
else:
    print(
        "\nCould not calculate linear regression: it requires at least "
        "two paired values and variation in the x-values."
    )

fig.tight_layout()
fig.savefig(regression_filename, dpi=200, bbox_inches="tight")
plt.close(fig)
# this saves the linear regression plot to a file within the repository 
print(f"Regression plot saved: {regression_filename}")
```

## Conclusions:

Our analysis aim was to determine whether the change in fat fraction percentage in the transversus abdominis (TA) had a similar sensitivity level to the functional test scores, specifically the patient's ability to walk or run 10 meters. We started by interpolating the data, using the minimum lag time between testing times as the target value. This created a new data set where the lag times between testing were the same. 

In order to analyze the patient condition as comprehensively as possible we combined the fat fraction score in the left and right TA at each time point. The t-test p value = 0.025, 0.0091 and 0.0070 (baseline values, t1 values, t2 values), which is greater than the commonly accepted significance value of 0.05. Therefore we can conclude there is no significant difference between the FSHD progression in the left and right TA. This conclusion compliments our hypothesis, as there has been no studies that have found FSHD progresses differently on the left or right side of the body. 

To calculate the sensitivity of the two measures we used an Standardarized Response Mean (SRM) test. Sensitivity and responsiveness are used interchangeably in regards to SRM results. They both describe how well a measure can represent change. SRM testing has been used by other researchers concerning this exact topic, attempting to find and justify new measures in clinical trials. A paper by Cornett et al. published in 2020, reported, “an SRM >0.8 is considered to indicate large responsiveness, 0.5–0.8 moderate, and 0.2–<0.5 low”. Their research was attempting to find the most sensitive subset of patients to analyze for clinical trials on neurological disease Charcot-Marie-Tooth (CMT). When comparing the SRM values we found in our data set to the ranges Cornett et al. proposes we can conclude the following. 

The responsiveness of fat fraction and the run walk test are not similar. Therefore, we must accept our null hypothesis. Fat fraction cannot properly represent the progression of FSHD specifically a Walk-Run functional Test. In other words, a clinical study that examined the fat fraction progression could not argue that the results would translate to a patient's function expression of the gene. However the change from T1 to T2 was not very “responsive” and had a low SRM score (SRM = 0.47173005899113407). ## Limitations and Future Work:

Limitations: 
One limitation of our study is the small sample size. The original study had 34 patients, but missing data reduced our analysis to 24 patients. A smaller sample size may limit the statistical power to detect meaningful differences between the left and right TA fat fraction and between changes in fat fraction and functional performance. Additionally, determining when disease progression begins can be quite challenging because there are only three measurement time points (baseline, T1, T2), which means short-term changes will not be identified as easily. Furthermore, the 10-meter walk/run test measures overall mobility and functional performance rather than TA function alone. The individual’s level of performance can be affected by fatigue, balance, and the overcompensation or weakness of other muscle groups to complete the test. Moreover, how one person with FSHD completes the test may look entirely different than that of another person. 

Ethical Implications: 
Patient privacy and confidentiality are very important ethical considerations to maintain when handling sensitive genetic and clinical information. Patient data must be securely stored in accordance with privacy requirements. Additionally, clinical trial outcomes can affect a patient’s ability to access future treatments. If sensitive endpoints are identified, then researchers can discover treatment benefits, but if inaccurate endpoints are identified, effective treatments can be overlooked causing ineffective treatments to appear beneficial. Researchers should also be aware of the financial burden, time commitment, and invasiveness of additional testing (MRI scans and biopsies).

Future Work: 
Future analyses should include a larger sample size to make the findings more generalizable to patients with FSHD. Researchers could also examine muscle biopsies from other muscle groups rather than solely focusing on the tibialis anterior to improve understanding of how fat infiltration and progression of the disease vary across the entire body. Future studies should also investigate whether early changes in fat fraction can predict long-term functional decline. Ultimately, future work should always prioritize patient safety, informed consent, and equitable representation.
