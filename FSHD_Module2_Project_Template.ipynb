## Project Goal:
As a team our goal is to determine if there are physiological or functional measures that can be used to measure facioscapulohumeral muscular dystrophy (FSHD) progression, beyond a traditional severity score. If measures are found that are proven to be accurate measurements of disease progression, we argue they should be used in clinical trials as additional endpoints. These additional endpoints could increase the accuracy of clinical trials by providing additional chances for trial success. Optimistically, more medications will pass the clinical trial phase and will be able to be put into market. As a team we are specifically focused on comparing the sensitivity of fat fraction in the TA and the functionality scores in the 10 meter walk run test. 

Our question is: 
Which measure is more sensitive to disease progression: the fat fraction in the left and right TA (tibialis anterior) or the score on the 10 meter walk run test (WR)? Furthermore, can a change in fat fraction in the TA in the first 12 months predict the patient's performance in the 10 meter WR test over the course of 2 years? 

By examining the fat fraction score in conjunction with a functionality test we are examining which is a better predictor of disease severity, and if the two measures impact each other. These results would provide vital knowledge that could be applied to designing clinical trials. We are specially looking at fat fraction in the TA because it is an easily accessible muscle to biopsy and commonly damaged over the course of FSHD progression. Furthermore, after conducting preliminary analysis with AI engine Claude, the fat fraction percent increase from baseline on the left side was 12.7% and on the right side was 10.6%, the two largest increases across all the measures in the provided data set. Therefore, we know the change in fat fraction has a similar trend as change in disease severity. These results provide justification for investigating fat fraction specifically. 
The TA contributes to an individual's ability to lift their foot off the ground, therefore our hypothesis is that as the fat fraction in the TA increases, strength in the TA decreases and the walk run score will decrease as well. We are examining specifically the walk and run test because it is the last stage of mobility before a patient is required to use a wheelchair. Therefore, the walk run test has the longest timespan and therefore possible data to track disease progression. 


## Disease Background: 
## Prevalence & Incidence

FSHD is the third most common inherited muscular dystrophies. Estimated global prevalence ranges between 1 in 7,500 and 1 in 20,000 individuals (or about 4 to 12 cases per 100,000 individuals). In the United States, total prevalent cases are estimated to be between 75,000 and 80,000 affected individuals. FSHD may be underestimated and milder cases can remain undiagnosed. The annual incidence rate is estimated to be 1 in 20,000 live births. Different regional/population studies can have varied recorded prevalence rates. For example, in the Netherlands, estimates of prevalence have a higher rate, whereas regions like China report lower prevalence rates due to unequal diagnostic access and underdiagnosis.

## Symptoms

Most symptoms manifest during adolescence of young adulthood, usually before twenty years of age. However, onset can span from early infancy to later adult life. The clinical feature of FSHD is an asymmetric pattern of progressive skeletal muscle weakness that begins in the upper body. The first sign is typically facial muscle weakness, which then shifts to weakness in the shoulder blades, scapular fixators, upper arms. The shoulder blades protrude outward to display “scapular winging.” Common complications include breathing problems, foot drop, lumbar lordosis from weak abdominal muscles, retinal damage, and hearing loss. These motor symptoms heavily impact the ability to complete routine daily tasks. Additionally, musculoskeletal pain and fatigue is common.
Sources: https://www.mda.org/disease/facioscapulohumeral-muscular-dystrophy/signs-and-symptoms

## Diagnosis

FSHD diagnosis begins with a clinical evaluation that examines muscle weakness, symptom onset, family history, and the distribution of affected muscles. Physicians look for physical signs such as asymmetric facial weakness, shoulder girdle atrophy, and scapular winging during physical examination. Genetic testing is the primary method used to confirm the diagnosis and distinguish FSHD type 1 from FSHD type 2. FSHD type 1, which accounts for approximately 95% of cases, is associated with contraction of the D4Z4 repeat region on chromosome 4, while FSHD type 2 (5% of cases) involves abnormal epigenetic regulation of this region. Additional tests, including creatine kinase blood testing, muscle MRI, and muscle biopsy may help evaluate muscle damage or rule out other conditions. Muscle MRIs are used to detect early muscular inflammation and fatty replacement, while creatine kinase levels remain normal or slightly elevated. Biopsy is now done only when FSHD is suspected but genetic testing does not confirm it. 
Sources: https://www.fshdsociety.org/living-with-fshd/understanding-fshd/diagnosis/
https://www.ncbi.nlm.nih.gov/books/NBK1443/

## Standard of Care Treatments (& reimbursement)

Currently, there are no FDA-approved disease-modifying therapies or medication to cure or halt the progression of FSHD, leaving clinical care strictly focused on supportive symptom management and improving quality of life. Multidisciplinary care teams emphasize physical therapy and individualized exercise programs to preserve existing muscle function, maintain joint range of motion, flexibility, strength, and prevent contractures. Orthotic devices such as wheelchairs or ankle-foot orthoses (AFOs) are regularly prescribed to assist patients experiencing foot drop and improve walking safety. Some patients can undergo scapulopexy (scapular fixation surgery), to stabilize the shoulder blades and improve arm reach and mobility. Managing chronic pain is a priority, often involving nonsteroidal anti-inflammatory drugs (NSAIDs) and neuropathic pain medications like gabapentin. Insurance reimbursement for physical therapy, assistive devices, surgery, pain management, and specialized genetic testing depends on medical necessity, the patient's insurance plan, and applicable coverage policies.  Ankle-foot orthoses are covered under the Medicare braces benefit when coverage criteria are met.

## Disease Progression & Prognosis

The natural progression of FSHD is typically slow, insidious, and variable, even among family members sharing the exact same genetic mutation. The severity can vary considerably between individuals as some symptoms may begin during childhood or adulthood. Some may experience mild weakness while others can develop significant physical disability. Muscle weakness generally follows a descending pattern, starting in the facial and shoulder muscles before gradually spreading downward toward the trunk, abdomen, and lower legs. Approximately 20% of affected individuals eventually require the use of a wheelchair or mobility scooter by age fifty to aid in performing everyday activities. The majority of individuals with classical FSHD have a normal life expectancy. However, if a patient presents with cardiac and respiratory complications or early-onset/infantile forms, they may require mechanical ventilation. Healthcare professionals adapt treatment throughout the patient's lifespan to meet each patient's needs.

## Biological Mechanisms (anatomy, organ physiology, cell & molecular physiology)

FSHD is caused by abnormal regulation of the DUX4 gene, which is located within the D4Z4 repeat region near the end of chromosome 4. In FSHD type 1, contraction of the D4Z4 repeat region makes it easier for DUX4 to become abnormally active in skeletal muscle cells. In FSHD type 2, mutations in genes involved in epigenetic regulation, particularly SMCHD1, can also reduce the normal silencing of this region. When DUX4 is expressed in muscle cells, it disrupts normal cellular processes and contributes to cellular stress, apoptotic cascades, muscle-cell damage, immune system activation, inflammation-related responses, and progressive muscle degeneration and muscle fiber necrosis.

## Clinical Trials/Next-gen therapies

Treatments that reduce DUX4 expression could potentially address an underlying cause of the disease rather than simply managing its symptoms. One approach involves antisense oligonucleotides and other RNA-targeting technologies designed to reduce DUX4 messenger RNA and limit production of the harmful protein. Del-brax, also known as AOC 1020, is an experimental therapy to reduce DUX4 messenger RNA in muscle cells. Del-brax has been evaluated in clinical trials for FSHD. Another drug, losmapimod, was studied in a Phase 3 clinical trial, but it did not show a significant benefit compared with a placebo, so development for FSHD was discontinued in 2024. Other research focuses on improving drug delivery to skeletal muscle, identifying biomarkers that measure disease activity, and developing more sensitive methods for evaluating muscle function. Although these therapies show potential, more research is needed to determine whether they are safe and effective for patients.

## Data Set:

Part of the data from the data set is from the study Regional and bilateral MRI and gene signatures in facioscapulohumeral dystrophy: implications for clinical trial design and mechanisms of disease progression by Wong et al. In this study they conducted the following test on 34 patients with FSHD to  produce their data. 

Clinical Severity Score (CSS): a 10-point scale rated by an examiner.

Quantitative myometry: a sensor that can track the force produced by a muscle in kg.

T1/T2 Dixon sequences: This is a form of MRI that separates water from fat. The results were used to calculate fat fraction. Fat fraction was computed both regionally (at the biopsy site) and across the whole TA muscle with an AI-based segmentation pipeline (Springbok Analytics).

STIR sequences: STIR is another form of MRI. It stands for Short Tau Inversion Recovery. It is a specific MRI that identifies edema and inflammation. The level of edema and inflammation is rated qualitatively. 

Muscle biopsy (needle biopsy of both the left and right TA):
Each muscle was biopsied, mostly within a few days of the MRI (up to four weeks for some subjects). 

Biopsy Analysis 

Histopathology: Microscopic analysis to identify changes due to disease progression.  

RNA sequencing: TRIzol extraction, 100 nt single-end Illumina sequencing 

DNA methylation analysis: bisulfite sequencing of the D4Z4 repeat region (4qA permissive allele, including 4qA161S vs. 4qA161L), plus SSLP haplotype analysis to confirm sample identity.

Other test that are mentioned in the data but not present in the supporting paper by Wong et al. are: 

Tug Time 
Timed Up and Go: time to stand, walk a short distance, turn and sit 

10 meter walk/run Test 
Time to cover 10 meters 

Ascend Test 
Time to ascend 4 stairs 

Descend Test 
Time to descend 4 stairs 


These test resulted in the following data points:

D4z4: a measure of the methylation of D4z4 

Dux4: Measuring the DUX4 signature. The expression of six DUX4-target genes (the DUX4 M6 score) represent DUX4 activity, since DUX4 itself is hard to detect

Lag Months (baseline, y1, y2):  the amount of time in months between each test. Units are months. 

Fat Fraction Right and Left: The fat fraction calculated from the MRIs in the left and right TA. Unit is a percentage. 

Fvar left and right: Not mentioned in paper so uncertain 

STIR: A measurement of inflammation and edema. 

Dorsiflexion strength: Collected with the quantitative myometry (kg) 

Tug Time: Not mentioned in paper. Likely measured in seconds. 

Walk Run Test: Not mentioned in paper. Likely measured in seconds. 

Ascend:  Not mentioned in paper. Likely measured in seconds. 

Descend:  Not mentioned in paper. Likely measured in seconds. 

Clinical Severity: Assessed by clinician on a scale of 0-10 





## MRI Data Analysis Methdology:

## Data Loading and Preprocessing
The MRI analysis was performed using Python, using NumPy for image-array processing, pandas for organizing metadata and results, and Matplotlib for visualizing the images and fat fraction measurements. The dataset contains paired water and fat MRI cross-sections of the TA from one participant during two visits. Each visit has eight slices or cross-sectional locations along the TA, which are not equally spaced. The slice_info.csv file included each slice’s physical distance and percentage location along the muscle. A metadata table links each slice to its water image, fat image and binary TA segmentation mask.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

DATA_DIR = Path("FSHD_TA_MRI_data_0020_TP1_TP2")
manifest = pd.read_csv(DATA_DIR / "slice_info.csv")
manifest
row = manifest[manifest["timepoint"] == "TP1"].iloc[3]  # representative mid-muscle slice

water = np.load(DATA_DIR / row["water_file"])
fat = np.load(DATA_DIR / row["fat_file"])
mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)

print("Water image shape:", water.shape)
print("Fat image shape:", fat.shape)
print("Mask shape:", mask.shape)
print("Number of muscle pixels:", mask.sum())
```

## Fat-Fraction Calculation
Fat fraction was calculated for each pixel as fat/(water+fat) * 100. To avoid division by zero and complications in the code, fat fraction was calculated where the sum of water and fat had to be greater than 0. The TA segmentation mask was able to distinguish between muscle pixels and surrounding tissue, which was excluded from the muscle-specific measurements. 


```python
denominator = water + fat
valid_pixels = denominator > 0

fat_fraction = np.divide(
    fat,
    denominator,
    out=np.zeros_like(fat, dtype=float),
    where=valid_pixels
) * 100

# Keep the whole supplied image region for the first analysis.
fat_fraction_full = fat_fraction.copy()

# Then apply the TA mask for the muscle-specific analysis.
fat_fraction_masked = fat_fraction.copy()
fat_fraction_masked[~mask] = np.nan

whole_image_values = fat_fraction_full[valid_pixels]
muscle_values = fat_fraction_masked[~np.isnan(fat_fraction_masked)]

print(f"Whole supplied image region — mean: {np.mean(whole_image_values):.1f}%")
print(f"Whole supplied image region — minimum: {np.min(whole_image_values):.1f}%")
print(f"Whole supplied image region — maximum: {np.max(whole_image_values):.1f}%")
print()
print(f"TA muscle only — mean: {np.mean(muscle_values):.1f}%")
print(f"TA muscle only — minimum: {np.min(muscle_values):.1f}%")
print(f"TA muscle only — maximum: {np.max(muscle_values):.1f}%")
```

## Image Visualization and Distribution
The water and fat images were displayed with the TA mask boundary to show which pixels belonged to the muscle. Fat fraction maps were generated using the 0-100% scale. 


```python
# Find the bounding box of the TA mask
y_pixels, x_pixels = np.where(mask)

y0, y1 = y_pixels.min(), y_pixels.max() + 1
x0, x1 = x_pixels.min(), x_pixels.max() + 1

# Crop the TA-specific fat-fraction map
fat_fraction_TA = fat_fraction_masked[y0:y1, x0:x1]

fig, axes = plt.subplots(1, 4, figsize=(15, 4))

# Water image with TA mask outline
axes[0].imshow(water, cmap="gray")
axes[0].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[0].set_title("Water image + TA mask")

# Fat image with TA mask outline
axes[1].imshow(fat, cmap="gray")
axes[1].contour(mask, levels=[0.5], colors="red", linewidths=2)
axes[1].set_title("Fat image + TA mask")

# Fat fraction for the whole supplied image region
whole_image = axes[2].imshow(
    fat_fraction_full,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[2].contour(mask, levels=[0.5], colors="cyan", linewidths=2)
axes[2].set_title("Fat fraction: whole region")

# Zoomed-in fat fraction for the TA only
ta_image = axes[3].imshow(
    fat_fraction_TA,
    cmap="plasma",
    vmin=0,
    vmax=100
)
axes[3].set_title("TA fat fraction")

for ax in axes:
    ax.axis("off")

# Both fat-fraction plots use the same 0–100% scale
fig.colorbar(
    ta_image,
    ax=axes[2:],
    fraction=0.046,
    pad=0.04,
    label="Fat fraction (%)"
)

plt.show()
```

Figure 1 Caption: Water and fat images, TA segmentation boundaries, and fat fraction maps. Reveals how the mask isolates the muscle for analysis.

```python
plt.figure(figsize=(7, 4))
plt.hist(whole_image_values, bins=30, alpha=0.55, label="Whole supplied region")
plt.hist(muscle_values, bins=30, alpha=0.65, label="TA muscle only")
plt.xlabel("Fat fraction (%)")
plt.ylabel("Number of pixels")
plt.title("Fat-fraction distributions")
plt.legend()
plt.show()
```

Figure 2 Caption: Histograms compare fat-fraction distributions across the whole image region and inside the TA mask. The mean alone may not capture the variability shown.

## Summary Statistics across muscle locations
The mean, median, and standard deviation of TA fat fraction were calculated for each slice. Additionally, the number of valid muscle pixels. The purpose of these measurements is to describe average muscle fat fraction, central value, variability among muscle pixels, and size of the analyzed region. The analysis was repeated for every sampled location in one visit.


```python
def calculate_fat_fraction(water, fat, mask):
    """Return a masked fat-fraction image in percent."""
    denominator = water + fat
    result = np.divide(
        fat,
        denominator,
        out=np.zeros_like(fat, dtype=float),
        where=denominator > 0
    ) * 100
    result[~mask] = np.nan
    return result


analysis_timepoint = "TP1"  # analyze one visit at a time
analysis_manifest = manifest[manifest["timepoint"] == analysis_timepoint]
results = []


for _, row in analysis_manifest.iterrows():
    water = np.load(DATA_DIR / row["water_file"])
    fat = np.load(DATA_DIR / row["fat_file"])
    mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
    ff = calculate_fat_fraction(water, fat, mask)


    results.append({
        "subject": row["subject"],
        "timepoint": row["timepoint"],
        "slice_id": row["slice_id"],
        "percent_along_muscle": row["percent_along_muscle"],
        "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
        "mean_fat_fraction_percent": np.nanmean(ff),
        "median_fat_fraction_percent": np.nanmedian(ff),
        "std_fat_fraction_percent": np.nanstd(ff),
        "n_muscle_pixels": int(mask.sum()),
    })


results_df = pd.DataFrame(results)
results_df
```

## Comparison between visits
To examine how fat fraction varied along the muscle, the mean fat fraction was plotted against physical distance. Since slices were not equally spaced, the supplied distances were used.


```python
def summarize_visit(timepoint):
    visit_results = []
    visit_manifest = manifest[manifest["timepoint"] == timepoint]


    for _, row in visit_manifest.iterrows():
        water = np.load(DATA_DIR / row["water_file"])
        fat = np.load(DATA_DIR / row["fat_file"])
        mask = np.load(DATA_DIR / row["mask_file"]).astype(bool)
        ff = calculate_fat_fraction(water, fat, mask)


        visit_results.append({
            "timepoint": timepoint,
            "slice_id": row["slice_id"],
            "percent_along_muscle": row["percent_along_muscle"],
            "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
            "mean_fat_fraction_percent": np.nanmean(ff),
        })


    return pd.DataFrame(visit_results)


profiles_df = pd.concat(
    [summarize_visit("TP1"), summarize_visit("TP2")],
    ignore_index=True
)


plt.figure(figsize=(7, 4))
for timepoint, group in profiles_df.groupby("timepoint"):
    group = group.sort_values("distance_from_muscle_start_mm")
    plt.plot(
        group["percent_along_muscle"],
        group["mean_fat_fraction_percent"],
        marker="o",
        label=timepoint
    )
plt.xlabel("Distance from muscle start (mm)")
plt.ylabel("Mean TA fat fraction (%)")
plt.title("Mean fat fraction along the tibialis anterior")
plt.legend(title="Visit")
plt.grid(True, alpha=0.3)
plt.show()
```

Figure 3 Caption: Mean TA Fat Fraction versus measured distance for TP1 and TP2.
